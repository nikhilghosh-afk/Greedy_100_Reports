```python
import lseg.data as ld
import pandas as pd
import csv
from tqdm import tqdm
ld.open_session()
print(f"{ld.__version__}")
```

```python
universe_ind = pd.read_csv('secs.csv')
tmp = universe_ind['PermID'].tolist()
g100_ind = [f"{p}@OrgPermID" for p in tmp]

data_fields = ['TR.CommonName','TR.TRBCActivity', 'TR.CompanyMarketCap']

df = ld.get_data(
    universe=g100_ind,
    fields=data_fields
)

df.to_csv("industries_sec.csv", index = False)
```

```python
# Creditors

universe_cred = pd.read_csv('secs.csv')
tmp = universe_cred['PermID'].tolist()
g100_cred = [f"{p}@OrgPermID" for p in tmp]


data_fields = [
    'TR.CommonName',
    'TR.LNDealFacilityId.DealId',
    'TR.LNAnnouncementDate',
    'TR.LNFinancialCloseDate',
    'TR.LNMaturityDate',
    'TR.LNTrancheType',
    'TR.LNUseOfProceeds',
    'TR.LNRepaymentType',
    'TR.LNIssuerUltParent',
    'TR.LNTotalFacilityAmount',
    'TR.LNManagerCount',
    'TR.LNManagerParentLong',
    'TR.LNManagerParentCode',
    'TR.LNManagerRole',
    'TR.LNCommitmentAmount'
]

df = ld.get_data(
    universe=g100_cred,
    fields=data_fields,
    parameters={'Curn' : 'USD'}
)

df.to_csv("creditors_sec.csv", index = False)
```

```python
# Owners

universe_own = pd.read_csv('secs.csv')

g100_own = universe_own['RIC'].dropna().tolist()

data_fields = [
    'TR.CommonName',
    'TR.InvestorFullName',
    'TR.SharesHeld',
    'TR.PctOfSharesOutHeld', 
    'TR.ClosePrice']

owners = ld.get_data(universe = g100_own,
                       fields = data_fields,
                    parameters = {'Curn': 'USD'})

owners.to_csv("owners_sec.csv", index = False)
```
